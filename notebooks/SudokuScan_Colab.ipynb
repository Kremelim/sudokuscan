## 2. One image

```python
!python sudoku_scan.py --input data/devset/images/dev_07.jpg --outdir out/

import cv2
import matplotlib.pyplot as plt

def show(*paths):
    figure, axes = plt.subplots(1, len(paths), figsize=(6 * len(paths), 6), squeeze=False)
    for axis, path in zip(axes[0], paths):
        axis.imshow(cv2.cvtColor(cv2.imread(path), cv2.COLOR_BGR2RGB))
        axis.set_title(path)
        axis.axis("off")
    plt.show()

show("out/dev_07_harris.png", "out/dev_07_debug.png")
show("out/dev_07_rect_A3.png", "out/dev_07_rect_A4.png", "out/dev_07_rect_H4.png")
!cat out/dev_07_grid.txt

# Bonus: solution (B1), solution on the photograph (B2), many-point homography (B3)
!cat out/dev_07_solution.txt
show("out/dev_07_ar.jpg", "out/dev_07_rect_HN.png")
```
